```python
import pandas as pd 
df = pd.read_csv('/kaggle/input/datasets/rohanrao/air-quality-data-in-india/city_day.csv')
print(df.head())
print(df.columns)
print(df.shape)
```

## 1.2 Missing Values & Basic Info

```python
print(df.isnull().sum())
print(df['City'].unique())
print(df['City'].nunique())
print(df['Date'].min(),df['Date'].max())
```

## 1.3 Date Parsing & Feature Extraction

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
```

```python
df.head()
```

## 1.4 Pollutant Ranges

```python
print(df['PM2.5'].max())
print(df['PM2.5'].min())
print(df['PM10'].max())
print(df['PM10'].min())
```

## 1.5 Season Mapping

Indian seasons mapped as:
- **Winter:** Dec, Jan, Feb
- **Spring:** Mar, Apr, May
- **Monsoon:** Jun–Sep
- **Autumn:** Oct, Nov

```python
def get_season(month):
    if(month<=2 or month == 12):
        return 'Winter'
    elif(3<=month<=5):
        return 'Spring'
    elif(6<=month<=9):
        return 'Monsoon'
    else:
        return 'Autumn'

df['Season'] = df['Month'].apply(get_season) 
```

## 1.6 Season-wise AQI Analysis

```python
print(df.groupby('Season')['AQI'].mean())
print(df.groupby('Season')['AQI'].min())
print(df.groupby('Season')['AQI'].max())
print(df.groupby('Season')['AQI'].median())
```

## 1.7 City × Season Pollution Breakdown

```python
pd.set_option('display.max_rows',None)
print(df.groupby(['Season','City'])['PM2.5'].mean())
print(df.groupby('City')['PM2.5'].mean().sort_values(ascending=False).head(20))
print(df.groupby('City')['AQI'].mean().sort_values(ascending=False).head())
```

## 1.8 COVID-19 Impact — Delhi (2019 vs 2020)

```python
delhi_covid = df[(df['City']=='Delhi') & (df['Year'] == 2020)]
print(delhi_covid.groupby('Month')['AQI'].mean())

delhi_2019 = df[(df['City'] == 'Delhi') & (df['Year'] == 2019)]
print(delhi_2019.groupby('Month')['AQI'].mean())
```
