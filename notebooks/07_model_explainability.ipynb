## 7.2 SHAP Analysis

Using `TreeExplainer` for efficient SHAP value computation on tree-based models.

```python
import shap

# Step 1: Explainer banao — model pass karo
explainer = shap.TreeExplainer(xgb_clf)

# Step 2: SHAP values nikalo — test data pe
shap_values = explainer.shap_values(X2_test)

# Step 3: Summary plot
shap.summary_plot(shap_values, X2_test)
shap.summary_plot(shap_values, X2_test, plot_type="bar")
```
